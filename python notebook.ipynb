```python
# 📦 Importing libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 🔍 Loading the data
df = pd.read_excel('Analytics_test_2.xlsx')

# 🎯 Overview
print("File shape:", df.shape)
print("Available columns:", df.columns.tolist())
print(df.head())
```

```python
# 1️⃣ Total registrations by visitor type
visitor_counts = df['Visitortype'].value_counts()
print("\nTotal by visitor type:\n", visitor_counts)

# 📊 Bar chart
plt.figure(figsize=(8,5))
sns.barplot(x=visitor_counts.index, y=visitor_counts.values, palette='viridis')
plt.title('Total Registrations by Visitor Type')
plt.xlabel('Visitor Type')
plt.ylabel('Total Registrations')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2️⃣ Top 5 countries represented
# Creating horizontal bar chart for top 5 countries represented
import matplotlib.pyplot as plt

# Data
countries = ['United States', 'Canada', 'Netherlands', 'Mexico', 'Japan']
registrations = [11437, 379, 111, 60, 54]

# Plotting
plt.style.use('seaborn-v0_8')
plt.figure(figsize=(10, 6))
bars = plt.barh(countries, registrations, color=plt.cm.viridis([0.1, 0.3, 0.5, 0.7, 0.9]))

# Labels and title
plt.xlabel('Total Registrations')
plt.ylabel('Country')
plt.title('Top 5 Countries Represented in the Event')
plt.gca().invert_yaxis()  # Highest value on top
plt.tight_layout()
```

```python
# 3️⃣ Percentage of selected answers to each demographic question
# 📊 Percentage of selected answers to each demographic question
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re

# Agrupando e calculando percentuais
demographic_distribution = df.groupby(['Demographic Question', 'Answer']).size().reset_index(name='Count')
demographic_distribution['Percent'] = demographic_distribution.groupby('Demographic Question')['Count'].transform(lambda x: x / x.sum() * 100)

# Lista de perguntas que devem usar gráfico horizontal
horizontal_questions = [
    "Q2. What best describes your job function?",
    "Q4. How did you hear about the show:",
    "Q7. I am looking for the following product groups: (check all that apply)",
    "Q8. I am involved/working in the following Application Areas: (check all that apply)"
]

# Função para limpar o título
def clean_question(text):
    return re.sub(r"\(.*?\)", "", text).strip()

# Gerando gráficos
for question in demographic_distribution['Demographic Question'].unique():
    subset = demographic_distribution[demographic_distribution['Demographic Question'] == question]
    print(f"\nDistribution for: {question}")
    print(subset[['Answer', 'Percent']].round(2))

    plt.figure(figsize=(10,6))
    title = clean_question(question)

    if question in horizontal_questions:
        sns.barplot(y='Answer', x='Percent', data=subset, palette='crest')
        plt.xlabel('Percentage (%)')
        plt.ylabel('Answer')
    else:
        sns.barplot(x='Answer', y='Percent', data=subset, palette='magma')
        plt.ylabel('Percentage (%)')
        plt.xlabel('Answer')
        plt.xticks(rotation=45)

    plt.title(f'Answer Distribution: {title}')
    plt.tight_layout()
    plt.show()
```

```python
# 💡 Additional insights

# 🔹 Geographic diversity
country_counts = df['Country'].value_counts()
print("\nTotal number of countries represented:", country_counts.count())

# 🔹 Participation by country and visitor type
pivot_country_visitor = pd.crosstab(df['Country'], df['Visitortype'])
print("\nDistribution by country and visitor type:\n", pivot_country_visitor.head())

# 🔹 Most answered demographic questions
question_counts = df['Demographic Question'].value_counts()
print("\nMost popular demographic questions:\n", question_counts)

# 🔹 Correlation between visitor type and demographic answers
pivot_visitor_demo = pd.crosstab(df['Visitortype'], df['Answer'], normalize='index') * 100
print("\nDemographic answer distribution by visitor type (%):\n", pivot_visitor_demo.round(1))
```
